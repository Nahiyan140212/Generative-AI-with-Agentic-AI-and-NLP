# Text Cleaning: Punctuation and White Space
**What it is:** Getting rid of messy stuff in text like extra spaces, tabs, or punctuation (e.g., commas, periods) that don’t add meaning.  
**Why it matters:** Makes text simpler and consistent for analysis, so the computer doesn’t get confused by random symbols or gaps.  

## Text Tokenization
**What it is:** Breaking text into smaller pieces, like splitting a sentence into words or phrases (e.g., `"I love coding"` → `["I", "love", "coding"]`).  
**Why it matters:** Turns text into bite-sized chunks that machines can process one by one.  

## Remove Stop Words
**What it is:** Filtering out common words like "the," "is," or "and" that don’t carry much meaning.  
**Why it matters:** Reduces noise, so the focus stays on important words that actually tell you something.  

## Lemmatization and Stemming
**What it is:** Simplifying words to their root form.  
- **Lemmatization** uses meaning (e.g., `"running"` → `"run"`).  
- **Stemming** just chops endings (e.g., `"running"` → `"runn"`).  

**Why it matters:** Groups similar words together, so `"runs," "running,"` and `"ran"` don’t get treated as totally different things.  

## Normalization and Text Segmentation
**What it is:**  
- **Normalization** makes text uniform (e.g., lowercase everything, fix typos).  
- **Segmentation** splits text into meaningful parts like sentences or paragraphs.  

**Why it matters:** Keeps everything consistent and organized for better processing.  

## Parts of Speech (POS) Tagging
**What it is:** Labeling each word with its role, like noun, verb, or adjective (e.g., `"Dog runs"` → `"Dog (noun) runs (verb)"`).  
**Why it matters:** Helps understand sentence structure and word functions.  

## Named Entity Recognition (NER)
**What it is:** Spotting and classifying names in text, like:  
- **People** (e.g., `"Elon Musk"`),  
- **Places** (`"New York"`),  
- **Organizations** (`"xAI"`).  

**Why it matters:** Pulls out key info for tasks like info extraction or question answering.  

## One-Hot Encoding
**What it is:** Turning words into a binary list where each word gets its own slot (e.g., `"cat"` → `[1, 0, 0]`, `"dog"` → `[0, 1, 0]`).  
**Why it matters:** Simple way to turn words into numbers for machines, but it’s basic and doesn’t show relationships.  

## Bag of Words (BoW)
**What it is:** Counting how often words appear in a document, ignoring order (e.g.,  
`"I like to code, I code"` → `{I: 2, like: 1, to: 1, code: 2}`).  
**Why it matters:** Quick way to represent text as numbers, but misses context or word placement.  

## TF-IDF (Term Frequency-Inverse Document Frequency)
**What it is:** Scoring words based on how common they are in one document but rare across others (e.g., `"quantum"` might score high in a physics paper).  
**Why it matters:** Highlights unique, important words instead of generic ones like `"the."`  

## Word2Vec
**What it is:** A method to turn words into vectors (number lists) where similar meanings are close together (e.g., `"king"` and `"queen"` have similar vectors).  
**Why it matters:** Captures word relationships and meaning, way better than BoW or One-Hot.  

## Word Embedding
**What it is:** A general term for mapping words to vectors in a way that preserves meaning (**Word2Vec is one type**).  
**Why it matters:** Gives machines a deeper understanding of language for tasks like translation or sentiment analysis.
