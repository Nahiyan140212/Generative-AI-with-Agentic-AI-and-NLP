### Resampling audio files
The sampling rate of an audio file determines the resolution. The higher the sampling rate, the higher the resolution which provides more detail about the sound wave itself.

When performing ML tasks it is important to ensure each file has the same sampling rate. This will maintain consistency and prepare the audio files based on what the model expects regarding number of data points per audio file.

The dataset, audio_file, and Audio from the datasets library are already loaded for you.

### Instructions

* Save the old sampling rate, for reference, as old_sampling_rate.
* Resample the audio column to a new rate of 16,000 kHz and save to audio_file.
* Compare the old and new sampling rates.

```python
from datasets import load_dataset, Audio

# Load an example audio dataset
audio_file = load_dataset("hf-internal-testing/librispeech_asr_dummy", split="validation")

# Save the old sampling rate (before resampling)
old_sampling_rate = audio_file[1]["audio"]["sampling_rate"]

# Resample the audio files to 16 kHz
audio_file = audio_file.cast_column("audio", Audio(sampling_rate=16_000))

# Reload the new sampling rate correctly
new_sampling_rate = audio_file[1]["audio"]["sampling_rate"]

# Compare the old and new sampling rates
print("Old sampling rate:", old_sampling_rate)
print("New sampling rate:", new_sampling_rate)
```

### Instantiating an ASR pipeline

You've been tasked with generating text from a dataset of audio files. Accuracy is important, so you need to make sure you choose the best model for automatic speech recognition. You also don't have the time to train your own model.

Compare the predictions between the Wav2Vec and Whisper models by instantiating two pipelines for automatic speech recognition. You want to test out the functionality, so you should try it out on one example first.

pipeline from the transformers package is already loaded for you. Likewise, the dataset has already been loaded for you, resampled, and saved as english. One audio file with it's associated metadata is saved as example.

### Instructions

* Instantiate the first automatic speech recognition pipeline for the "wav2vec2" model from Meta.
* Predict the text from the example audio.
* Repeat these two steps for the "whisper-tiny" model from OpenAI in order to compare the predictions.

```python
from transformers import pipeline
from datasets import load_dataset
import torchaudio
import torch

# Load a sample dataset from Hugging Face
dataset = load_dataset("PolyAI/minds14", name="en-US", split="train", trust_remote_code=True)
example = dataset[0]  # Pick the first sample

# Extract audio metadata
audio_info = example["audio"]
audio_array = audio_info["array"]
sampling_rate = audio_info["sampling_rate"]

print(f"Original Sampling Rate: {sampling_rate} Hz")

# Convert audio to float32 and resample if needed
audio_tensor = torch.tensor(audio_array, dtype=torch.float32)  # 🔹 Convert to Float32

if sampling_rate != 16000:
    print("Resampling audio to 16kHz...")
    resampler = torchaudio.transforms.Resample(orig_freq=sampling_rate, new_freq=16000)
    audio_tensor = resampler(audio_tensor)

# Convert back to NumPy array
audio_array = audio_tensor.numpy()

# Load Meta's Wav2Vec2 model
meta_asr = pipeline(task="automatic-speech-recognition", model="facebook/wav2vec2-base-960h")

# Predict the text using Wav2Vec2
meta_pred = meta_asr(audio_array)["text"].lower()

# Load OpenAI's Whisper model (using a larger model for better accuracy)
open_asr = pipeline("automatic-speech-recognition", model="openai/whisper-base")

# Predict the text using Whisper
open_pred = open_asr(audio_array)["text"].lower()

# Print the predictions
print("META (Wav2Vec2):", meta_pred)
print("OPENAI (Whisper):", open_pred)
```

## Iterating over a dataset
You were able test the functionality and understand the performance on one example from the dataset. Now, let's evaluate the models over the first 100 audio files to make a final decision about which is best for this dataset.

In order to do this efficiently, you can create a function that will iterate over the rows of the dataset and yield a set of audio and true sentence pairs on each iteration.

The dataset, english, ASR models - meta_asr and open_asr - and pandas have all been loaded for you.

### Instructions 

* Create the data() function to iterate over the first 3 rows of the dataset.
* Within data(), predict the text for each audio file using both the meta_asr and open_asr pipelines.
* Append the results as a dictionary in the output list.

```python
import pandas as pd
import torch
import torchaudio
from transformers import pipeline
from datasets import load_dataset
from evaluate import load  # ✅ Use `evaluate` instead of deprecated `load_metric`

# 🔹 Load the dataset (English dataset from PolyAI Minds14)
dataset = load_dataset("PolyAI/minds14", name="en-US", split="train", trust_remote_code=True)

# 🔹 Check the available keys
print(dataset.features)

# ✅ Define the correct key for ground-truth text
true_text_key = "english_transcription"  # 🔹 Adjust if necessary

# ✅ Extract the first few samples for testing
def data(n=3):
    for i in range(n):
        yield dataset[i]["audio"]["array"], dataset[i][true_text_key].lower()

# ✅ Load ASR models
meta_asr = pipeline(task="automatic-speech-recognition", model="facebook/wav2vec2-base-960h")
open_asr = pipeline("automatic-speech-recognition", model="openai/whisper-base")  # 🔹 Use a larger model for better results

# ✅ Predict text from the audio samples
output = []
for audio, sentence in data():
    meta_pred = meta_asr(audio)["text"].lower()
    open_pred = open_asr(audio)["text"].lower()
    # Append to output list
    output.append({"sentence": sentence, "metaPred": meta_pred, "openPred": open_pred})

# ✅ Convert to a Pandas DataFrame
output_df = pd.DataFrame(output)
```

```python
from datasets import load_dataset
from transformers import AutoModelForSequenceClassification, AutoTokenizer, TrainingArguments, Trainer

# ✅ Load sentiment analysis dataset (IMDB movie reviews)
dataset = load_dataset("imdb")

# ✅ Split into training (90%) and testing (10%)
dataset = dataset["train"].train_test_split(test_size=0.1)
train_data = dataset["train"]
test_data = dataset["test"]

# ✅ Define text & label keys
text_key = "text"      # 🔹 Text of the review
label_key = "label"    # 🔹 Sentiment (0 = negative, 1 = positive)

# ✅ Get number of unique labels (2 for binary sentiment analysis)
num_labels = len(set(train_data[label_key]))

# ✅ Load DistilBERT model for sentiment analysis
model_name = "distilbert-base-uncased"
model = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=num_labels,  # 🔹 Match dataset classes (2)
    ignore_mismatched_sizes=True  # ✅ Avoid size mismatch error
)

# ✅ Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)

# ✅ Tokenize the dataset
def preprocess_function(batch):
    encoding = tokenizer(batch[text_key], padding=True, truncation=True)
    encoding["labels"] = batch[label_key]  # 🔹 Add labels for training
    return encoding

train_data = train_data.map(preprocess_function, batched=True)
test_data = test_data.map(preprocess_function, batched=True)

# ✅ Define training arguments
training_args = TrainingArguments(
    output_dir="./sentiment_results",
    evaluation_strategy="epoch",
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,  # 🔹 Increase for better performance
    save_strategy="epoch",
    logging_dir="./logs",
    logging_steps=10
)

# ✅ Create Trainer
trainer = Trainer(
    model=model, 
    args=training_args, 
    train_dataset=train_data,
    eval_dataset=test_data
)

# ✅ Start training
trainer.train()

# ✅ Save the fine-tuned model
trainer.save_model("./fine_tuned_sentiment_model")
tokenizer.save_pretrained("./fine_tuned_sentiment_model")
```

### Using semantic search
The similarity, or closeness, between a query and the other sentences, or documents, is the foundation for semantic search. This is a search method which takes into account context and intent of the query. Similarity measures, such as cosine similarity, are used to quantify the distance between the query and each sentence within the dimensional space. Results of a search are based on the closest sentences to the query.

You will use semantic search to return the top two Reddit threads relevant to the user query, "I need a desktop book reader for Mac".

```python
from sentence_transformers import SentenceTransformer, util

# ✅ Load the Sentence Transformer model (optimized for retrieval tasks)
embedder = SentenceTransformer("multi-qa-MiniLM-L6-cos-v1")  # 🔹 A great model for semantic search

# ✅ Example dataset: Reddit threads (simulated)
reddit_threads = [
    "Does anyone know a good eBook reader for macOS?",
    "Looking for a Kindle alternative for Mac.",
    "Where to buy a MacBook at a discount?",
    "Best free PDF reader for Mac?",
    "Mac or Windows? Which one is better for developers?",
    "How to install Windows on Mac?",
    "Best desktop application for reading books on macOS?"
]

# ✅ Generate embeddings for Reddit threads
thread_embeddings = embedder.encode(reddit_threads, convert_to_tensor=True)

# ✅ User query for semantic search
query = "I need a desktop book reader for Mac"

# ✅ Generate embedding for the query
query_embedding = embedder.encode(query, convert_to_tensor=True)  # No need for [query][0]

# ✅ Perform semantic search (return top 2 matches)
hits = util.semantic_search(query_embedding, thread_embeddings, top_k=2)

# ✅ Print the top two most relevant Reddit threads
print("\n🔍 Top 2 Relevant Reddit Threads:")
for hit in hits[0]:  # The first (and only) query in `hits`
    print(f"🔹 {reddit_threads[hit['corpus_id']]} (Score: {hit['score']:.4f})")
```
